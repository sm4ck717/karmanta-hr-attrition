```python
df = df[(df["Age"].notna()) & (df["Age"] != 999)]
```

```python
df["Salary"] = (
    df["Salary"]
    .astype(str)
    .str.replace("'", "", regex=False)
)

df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
```

```python
df['Salary'].dtype
```

```python
df['Salary'].unique()
```

```python
df['Salary'].describe()
```

```python
df['Total_working_years_experience'].head(20).tolist()
```

```python
df["Total_working_years_experience"] = (
    df["Total_working_years_experience"]
    .astype(str)
    .str.replace("O", "0", regex=False)
)
```

```python
df["Total_working_years_experience"] = (
    df["Total_working_years_experience"]
    .str.replace(r"[^\d]", "", regex=True)
)
```

```python
df["Total_working_years_experience"] = pd.to_numeric(
    df["Total_working_years_experience"], errors="coerce"
)
```

```python
df['Total_working_years_experience'].dtype
```

```python
df['Total_working_years_experience'].unique()
```

```python
num_cols = df.select_dtypes(include=["int64", "float64"]).columns

df[num_cols].describe()
```

```python
df[num_cols].describe().T
```

```python
cat_cols = df.select_dtypes(include=["object"]).columns

for col in cat_cols:
    print("\n", col)
    print(df[col].value_counts())
```

```python
for col in cat_cols:
    print(f"\n==== {col} ====")
    display(df[col].value_counts(dropna=False))
    display((df[col].value_counts(normalize=True) * 100).round(2))
```

```python
print("Numeric:", len(num_cols))
print("Categorical:", len(cat_cols))
```

```python
mask = df["No_of_years_worked_at_current_company"] > df["Total_working_years_experience"]

df.loc[mask, "Total_working_years_experience"] = df.loc[
    mask, "No_of_years_worked_at_current_company"
]
```

```python
(df["No_of_years_worked_at_current_company"] >
 df["Total_working_years_experience"]).sum()
```

```python
df["External_experience"] = (
    df["Total_working_years_experience"]
    - df["No_of_years_worked_at_current_company"]
)
```
