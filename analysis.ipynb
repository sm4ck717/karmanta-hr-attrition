## Demographic Analysis & Overall Stats

```python
df["Attrition"].value_counts(normalize=True)["Yes"] * 100
```

```python
pd.DataFrame(df["Attrition"].value_counts(normalize=True) * 100)
```

```python
pd.crosstab(df["Native_Place"], df["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(df["Gender"], df["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(df["Marital_Status"], df["Attrition"], normalize="index") * 100
```

## Burnout & Education Impact

```python
pd.crosstab(df["Overtime"], df["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(df["Education"], df["Attrition"], normalize="index") * 100
```

## Growth Stagnation

```python
pd.DataFrame(df.groupby("Attrition")["Years_since_last_promotion"].mean())
```

```python
pd.DataFrame(df.groupby("Attrition")["No_of_years_in_current_role"].mean())
```

## Satisfaction Metrics

## Pay & Compensation

```python
pd.DataFrame(df.groupby("Attrition")["Salary"].mean())
```

```python
pd.DataFrame(df.groupby("Attrition")["Salary_Hike_in_percent"].mean())
```

## Career Stage Analysis

```python
df["age_band"] = pd.cut(
    df["Age"],
    bins=[18, 25, 35, 45, 55, 65]
)

pd.crosstab(df["age_band"], df["Attrition"], normalize="index") * 100
```

## Structural Metrics

```python
pd.crosstab(df["Job_Level"], df["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(df["Sub-department"], df["Attrition"], normalize="index") * 100
```

## Distance & Travel Impact

```python
df["distance_band"] = pd.cut(
    df["Distance_From_Home"],
    bins=[0,10,20,30,40,50]
)

pd.crosstab(df["distance_band"], df["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(df["Business_Travel"], df["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(
    pd.cut(df["External_experience"], bins=[0,2,5,10,20]),
    df["Attrition"],
    normalize="index"
) * 100
```
